## Preprocess Image Data

### Subtask:
Apply necessary preprocessing to the image data, such as resizing to a uniform dimension, converting to numerical arrays, and normalizing pixel values. Also, prepare the annotations (labels) for model training.


**Reasoning**:
I will import the necessary libraries (`cv2` and `numpy`) and define the `preprocess_image` function as per the instructions. This function will read, resize, and normalize images. Afterwards, I'll apply this function to the training, testing, and validation image paths and convert the resulting lists of images and labels into NumPy arrays for model training.



```python
import cv2
import numpy as np

# 2. Define a consistent image size
IMAGE_SIZE = (224, 224) # Width, Height

# 3. Create a function to preprocess images
def preprocess_image(image_path):
    try:
        img = cv2.imread(image_path)
        if img is None:
            print(f"Warning: Could not read image at {image_path}. Skipping.")
            return None
        img = cv2.resize(img, IMAGE_SIZE)
        img = img / 255.0  # Normalize pixel values to [0, 1]
        return img
    except Exception as e:
        print(f"Error processing image {image_path}: {e}")
        return None

# 4. Apply the preprocess_image function to each image path
X_train_images = []
for path in train_image_paths:
    processed_img = preprocess_image(path)
    if processed_img is not None:
        X_train_images.append(processed_img)

X_test_images = []
for path in test_image_paths:
    processed_img = preprocess_image(path)
    if processed_img is not None:
        X_test_images.append(processed_img)

X_valid_images = []
for path in valid_image_paths:
    processed_img = preprocess_image(path)
    if processed_img is not None:
        X_valid_images.append(processed_img)

# 5. Convert lists to NumPy arrays
X_train_images = np.array(X_train_images)
X_test_images = np.array(X_test_images)
X_valid_images = np.array(X_valid_images)

y_train_images = np.array(train_labels)
y_test_images = np.array(test_labels)
y_valid_images = np.array(valid_labels)

print(f"X_train_images shape: {X_train_images.shape}")
print(f"y_train_images shape: {y_train_images.shape}")
print(f"X_test_images shape: {X_test_images.shape}")
print(f"y_test_images shape: {y_test_images.shape}")
print(f"X_valid_images shape: {X_valid_images.shape}")
print(f"y_valid_images shape: {y_valid_images.shape}")
```

## Build Image Classification Model

### Subtask:
Construct a deep learning model, likely a Convolutional Neural Network (CNN), suitable for classifying the image data based on the provided annotations. This model will be designed to predict cancer risk from the images.


**Reasoning**:
I will import the necessary modules from `tensorflow.keras`, then define and compile a Sequential CNN model with `Conv2D`, `MaxPooling2D`, `Flatten`, and `Dense` layers, and finally print its summary.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Input

# 2. Initialize a Sequential model
model_cnn = Sequential()

# 3. Add an Input layer
# Assuming IMAGE_SIZE is defined and images are RGB (3 channels)
model_cnn.add(Input(shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3)))

# 4. Add several Conv2D and MaxPooling2D layers
model_cnn.add(Conv2D(32, (3, 3), activation='relu'))
model_cnn.add(MaxPooling2D((2, 2)))
model_cnn.add(Conv2D(64, (3, 3), activation='relu'))
model_cnn.add(MaxPooling2D((2, 2)))
model_cnn.add(Conv2D(128, (3, 3), activation='relu'))
model_cnn.add(MaxPooling2D((2, 2)))

# 5. Flatten the output
model_cnn.add(Flatten())

# 6. Add one or more Dense (fully connected) layers
model_cnn.add(Dense(128, activation='relu'))

# 7. Add a final Dense output layer for binary classification
model_cnn.add(Dense(1, activation='sigmoid'))

# 8. Compile the model
model_cnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 9. Print the model summary
model_cnn.summary()
```

**Reasoning**:
Now that the CNN model has been constructed and compiled, the next step is to train it using the preprocessed training and validation image data and their corresponding labels.



```python
history = model_cnn.fit(
    X_train_images,
    y_train_images,
    epochs=3,  # You can adjust the number of epochs
    validation_data=(X_valid_images, y_valid_images),
    batch_size=64 # You can adjust the batch size
)

print("Model training complete.")
```
